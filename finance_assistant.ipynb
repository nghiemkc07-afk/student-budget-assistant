# Six-Step Problem-Solving Method


# Step 1: Understand the Problem

**Project Name:** Student Budget Assistant

**Problem:** University students often have limited income and may struggle to manage their monthly expenses. Without a clear understanding of their spending habits, they may overspend and find it difficult to save money.

**Target Users:** University students, particularly those managing their own finances while studying and working part-time.

**Proposed Solution:** The Student Budget Assistant is a Python-based application designed to help students manage their personal finances. Users can enter their monthly income and expenses across different categories, such as food, transport, rent and shopping. The application will calculate total expenses, remaining balance and spending percentages. It will also integrate a Gemini-powered chatbot to provide personalised financial guidance based on the user's data.

**Project Goal:** To help students understand their spending patterns, make informed financial decisions and improve their budgeting habits.

## Step 2: Identify Inputs and Outputs

**Inputs:**

The application will require the following information from the user:

- **Monthly Income:** The total amount of money the student receives each month (AUD).
- **Food Expenses:** Monthly spending on groceries, meals and drinks.
- **Transport Expenses:** Monthly spending on public transport, fuel or other transportation.
- **Rent Expenses:** Monthly accommodation costs.
- **Shopping Expenses:** Monthly spending on clothes, personal items and other purchases.
- **Other Expenses:** Any additional monthly expenses not included in the categories above.
- **User Questions:** Finance-related questions entered into the AI chatbot.

**Outputs:**

Based on the user's inputs, the application will provide:

1. **Total Monthly Expenses:** The sum of all expense categories.
2. **Remaining Balance:** Monthly income minus total monthly expenses.
3. **Spending Breakdown:** The percentage of total expenses allocated to each category.
4. **Budget Status:** An indication of whether the user is spending within their monthly income or exceeding it.
5. **Personalised AI Responses:** Financial guidance generated by Gemini AI based on the user's actual income, expenses and questions.

**Input Validation:**

The application will check that income and expense values are valid, non-negative numbers. Invalid inputs will produce a clear error message rather than causing the application to crash.

## Step 4: Write Pseudocode

The following pseudocode outlines the logic of the Student Budget Assistant before converting it into Python.

### Part 1: Collect and Validate User Inputs

```text
START

INPUT monthly_income
INPUT food_expenses
INPUT transport_expenses
INPUT rent_expenses
INPUT shopping_expenses
INPUT other_expenses

IF any input is not a valid number OR is negative THEN
    DISPLAY "Invalid input. Please enter non-negative numbers."
    STOP

```

### Part 2: Calculate Monthly Budget

```text
SET total_expenses = food_expenses + transport_expenses
                   + rent_expenses + shopping_expenses
                   + other_expenses

SET remaining_balance = monthly_income - total_expenses

IF total_expenses > 0 THEN
    FOR each expense category
        CALCULATE percentage = (category_expense / total_expenses) * 100
    END FOR
ELSE
    SET all spending percentages = 0
END IF
```

### Part 3: Determine Budget Status

```text
IF total_expenses <= monthly_income THEN
    SET budget_status = "Within Budget"
ELSE
    SET budget_status = "Over Budget"
END IF
```

### Part 4: Display Results

```text
DISPLAY total_expenses
DISPLAY remaining_balance
DISPLAY spending percentages for each category
DISPLAY budget_status
```

### Part 5: AI Financial Assistant

```text
INPUT user_question

PREPARE financial_summary using the calculated results

SEND user_question and financial_summary to Gemini AI

IF the AI request is successful THEN
    DISPLAY personalised financial advice
ELSE
    DISPLAY "Unable to generate advice. Please try again."
END IF

END
```

## Step 5: Convert to Python

In this step, I will convert the pseudocode into Python functions. The program will validate user inputs, calculate total monthly expenses, determine the remaining balance, calculate spending percentages, and identify whether the user is within or over budget.

````python
def calculate_budget(income, expenses):
    # Validate income
    if not isinstance(income, (int, float)) or income < 0:
        raise ValueError("Income must be a non-negative number.")

    # Validate expense categories
    for category, amount in expenses.items():
        if not isinstance(amount, (int, float)) or amount < 0:
            raise ValueError("Expenses must be non-negative numbers.")

    # Calculate total expenses and remaining balance
    total_expenses = sum(expenses.values())
    remaining_balance = income - total_expenses

    # Calculate spending percentages
    percentages = {}

    for category, amount in expenses.items():
        if total_expenses > 0:
            percentages[category] = round(
                (amount / total_expenses) * 100, 2
            )
        else:
            percentages[category] = 0

    # Determine budget status
    if total_expenses <= income:
        budget_status = "Within Budget"
    else:
        budget_status = "Over Budget"

    return {
        "total_expenses": total_expenses,
        "remaining_balance": remaining_balance,
        "percentages": percentages,
        "budget_status": budget_status
    }
````

````python

income = 2000

expenses = {
    "Food": 400,
    "Transport": 300,
    "Rent": 700,
    "Shopping": 200,
    "Other": 100
}

result = calculate_budget(income, expenses)

print("Total Expenses:", result["total_expenses"])
print("Remaining Balance:", result["remaining_balance"])
print("Budget Status:", result["budget_status"])
print("Spending Percentages:")

for category, percentage in result["percentages"].items():
    print(category, ":", percentage, "%")
````

Normal case

````python

# Test 1: Normal monthly budget
test1 = calculate_budget(2000, {
    "Food": 400,
    "Transport": 300,
    "Rent": 700,
    "Shopping": 200,
    "Other": 100
})

assert test1["total_expenses"] == 1700
assert test1["remaining_balance"] == 300
assert test1["budget_status"] == "Within Budget"
assert test1["percentages"]["Food"] == 23.53

print("Test 1 Passed: Normal budget")


# Test 2: Overspending
test2 = calculate_budget(1000, {
    "Food": 500,
    "Rent": 800
})

assert test2["total_expenses"] == 1300
assert test2["remaining_balance"] == -300
assert test2["budget_status"] == "Over Budget"

print("Test 2 Passed: Overspending")


# Test 3: Zero expenses
test3 = calculate_budget(1500, {
    "Food": 0,
    "Transport": 0
})

assert test3["total_expenses"] == 0
assert test3["remaining_balance"] == 1500
assert test3["percentages"]["Food"] == 0

print("Test 3 Passed: Zero expenses")
````
